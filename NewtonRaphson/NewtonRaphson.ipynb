```python
def iterate(x: complex, count: int =0) -> complex:
    xn1 = (x - ((x**2+1)/(2*x))).round(2)

    if count == 16:
        return xn1
    
    return (iterate(xn1, count+1))
```

```python
x = np.round(np.linspace(-10, 10, 100, dtype=complex), 2)
y = np.round(np.linspace(-10j, 10j, 100, dtype=complex), 2)
inputs = x + y[:, np.newaxis]


#roots
r1, r2 = -1j, 1j
```

```python
outputs = iterate(inputs)
```

```python
map = np.zeros((100,100), dtype=int)
 
```

```python
diff1 = r1 - outputs
diff2 = r2 - outputs
```

```python
map[diff1==(0+0j)] = 1
map[diff2==(0+0j)] = 2
```

```python
fig, ax = plt.subplots()

plt.imshow(map)
ax.invert_yaxis()
plt.axis('off')
plt.show()
```
